```python
"""
Compare three recommenders on MovieLens Latest Small and explain their output.

Models:
  1. Popularity   - Bayesian weighted rating, same list for every user.
  2. Content      - TF-IDF on title + genres; user profile = mean of liked movies.
  3. SVD          - Truncated SVD on mean-centred user-item ratings (latent factors).

Evaluation: per-user 80/20 hold-out (seed 42). A held-out movie is relevant if rated >= 4.0.
Metrics: Precision@10, Recall@10, RMSE/MAE (SVD only), diversity of top-10 lists.
Explanations (XAI):
  - Content: which title/genre terms in the user's profile pushed each recommended movie up.
  - SVD: which genres the user rates above or below their own average.
"""
import re
import sys
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
from sklearn.metrics.pairwise import cosine_similarity

# DATA = sys.argv[1]          # path to the ml-latest-small folder
# OUT = sys.argv[2]           # output folder for CSV/PNG/
DATA = "ml-latest-small"   # folder containing ratings.csv and movies.csv
OUT = "."                  # where results and the chart are saved
SEED = 42
K = 10
N_USERS = 200               # users evaluated (first N with >=1 relevant test item)
LIKE = 4.0                  # relevance / liking threshold
N_FACTORS = 50

rng = np.random.default_rng(SEED)
ratings = pd.read_csv(f"{DATA}/ratings.csv")
movies = pd.read_csv(f"{DATA}/movies.csv")
movies["clean_title"] = movies["title"].str.replace(r"\s*\(\d{4}\)\s*$", "", regex=True)
movies["genres"] = movies["genres"].fillna("").str.replace("|", " ", regex=False)

# ---- per-user 80/20 hold-out -------------------------------------------------
train_parts, test_parts = [], []
for uid, grp in ratings.groupby("userId"):
    idx = rng.permutation(len(grp))
    n_test = max(1, int(round(0.2 * len(grp))))
    test_parts.append(grp.iloc[idx[:n_test]])
    train_parts.append(grp.iloc[idx[n_test:]])
train = pd.concat(train_parts)
test = pd.concat(test_parts)

user_ids = np.sort(ratings["userId"].unique())
movie_ids = np.sort(movies["movieId"].unique())
u_index = {u: i for i, u in enumerate(user_ids)}
m_index = {m: i for i, m in enumerate(movie_ids)}
n_u, n_m = len(user_ids), len(movie_ids)
title_of = dict(zip(movies["movieId"], movies["title"]))
genres_of = dict(zip(movies["movieId"], movies["genres"].str.split()))

# dense train matrix (610 x 9724 fits easily)
R_train = np.zeros((n_u, n_m))
for u, m, r in zip(train.userId, train.movieId, train.rating):
    R_train[u_index[u], m_index[m]] = r
seen = R_train > 0
user_mean = np.array([R_train[i][seen[i]].mean() if seen[i].any() else 0.0 for i in range(n_u)])

# ---- Model 1: popularity (Bayesian weighted rating, train only) --------------
g = train.groupby("movieId")["rating"].agg(["count", "mean"])
C = train["rating"].mean()
m_min = g["count"].quantile(0.9)
wr = (g["count"] / (g["count"] + m_min)) * g["mean"] + (m_min / (g["count"] + m_min)) * C
pop_scores = np.full(n_m, -np.inf)
for mid, s in wr.items():
    pop_scores[m_index[mid]] = s
pop_scores_all = np.tile(pop_scores, (n_u, 1))

# ---- Model 2: content-based (user profile from liked movies) -----------------
tfidf = TfidfVectorizer(stop_words="english", ngram_range=(1, 2))
docs = (movies.set_index("movieId").loc[movie_ids, "clean_title"] + " " +
        movies.set_index("movieId").loc[movie_ids, "genres"]).str.lower()
X = tfidf.fit_transform(docs.values)                     # n_m x vocab (sparse)
terms = np.array(tfidf.get_feature_names_out())
liked = (R_train >= LIKE).astype(float)
profiles = (liked @ X)                                    # n_u x vocab (sparse)
profiles = np.asarray(profiles)
norms = np.linalg.norm(profiles, axis=1, keepdims=True)
norms[norms == 0] = 1
profiles = profiles / norms
content_scores_all = profiles @ X.T.toarray()             # n_u x n_m

# ---- Model 3: SVD on mean-centred ratings -----------------------------------
centred = np.where(seen, R_train - user_mean[:, None], 0.0)
svd = TruncatedSVD(n_components=N_FACTORS, random_state=SEED)
U = svd.fit_transform(centred)
V = svd.components_
svd_pred_all = user_mean[:, None] + U @ V                 # n_u x n_m
svd_pred_all = np.clip(svd_pred_all.T, 0.5, 5.0).T

# ---- evaluation helpers ------------------------------------------------------
test_rel = {}
for u, m, r in zip(test.userId, test.movieId, test.rating):
    if r >= LIKE:
        test_rel.setdefault(u_index[u], set()).add(m_index[m])

def top_k(scores_row, seen_row, k=K):
    s = scores_row.copy()
    s[seen_row] = -np.inf
    return np.argsort(-s)[:k]

def diversity(items):
    sub = X[items]
    sim = cosine_similarity(sub)
    n = len(items)
    return 1 - (sim.sum() - n) / (n * (n - 1))

eval_users = [u for u in sorted(test_rel) if len(test_rel[u]) > 0][:N_USERS]
rows = []
lists = {"Popularity": {}, "Content-based": {}, "SVD": {}}
for name, scores_all in [("Popularity", pop_scores_all),
                         ("Content-based", content_scores_all),
                         ("SVD", svd_pred_all)]:
    precs, recs, divs = [], [], []
    for u in eval_users:
        rec = top_k(scores_all[u], seen[u])
        lists[name][u] = rec
        hits = len(set(rec) & test_rel[u])
        precs.append(hits / K)
        recs.append(hits / len(test_rel[u]))
        divs.append(diversity(rec))
    rows.append({"Model": name, "Users": len(eval_users),
                 "Precision@10": np.mean(precs), "Recall@10": np.mean(recs),
                 "Diversity": np.mean(divs)})

# RMSE / MAE for SVD on the held-out ratings
errs = []
for u, m, r in zip(test.userId, test.movieId, test.rating):
    errs.append(svd_pred_all[u_index[u], m_index[m]] - r)
errs = np.array(errs)
rmse, mae = np.sqrt(np.mean(errs ** 2)), np.mean(np.abs(errs))
results = pd.DataFrame(rows)
results.loc[results["Model"] == "SVD", "RMSE"] = rmse
results.loc[results["Model"] == "SVD", "MAE"] = mae
results.to_csv(f"{OUT}/results.csv", index=False)
print(results.round(4).to_string(index=False))

# ---- XAI: explain one user's recommendations --------------------------------
explain_user = eval_users[0]
uid = user_ids[explain_user]
print(f"\n=== Explanations for user {uid} ===")
liked_titles = movies.set_index("movieId").loc[
    [m for m, r in zip(train.movieId[train.userId == uid], train.rating[train.userId == uid]) if r >= LIKE][:8],
    "title"].tolist()
print("Some movies this user liked (train):", "; ".join(liked_titles))

# content: top profile terms and per-movie contribution
prof = profiles[explain_user]
top_terms = terms[np.argsort(-prof)[:10]]
print("Top profile terms:", ", ".join(top_terms))

# genre affinity for SVD explanation
user_rows = train[train.userId == uid].merge(movies, on="movieId")
genre_rows = []
for _, row in user_rows.iterrows():
    for gname in str(row["genres"]).split():
        genre_rows.append((gname, row["rating"] - user_mean[explain_user]))
affinity = pd.DataFrame(genre_rows, columns=["genre", "diff"]).groupby("genre")["diff"].mean()

rec_svd = lists["SVD"][explain_user][:5]
expl = []
for mi in rec_svd:
    mid = movie_ids[mi]
    gs = genres_of[mid]
    aff = [affinity.get(gn, np.nan) for gn in gs]
    aff = [a for a in aff if not np.isnan(a)]
    why = ", ".join(f"{gn} ({affinity[gn]:+.2f})" for gn in gs if gn in affinity.index) or "no genre history"
    expl.append({"Movie": title_of[mid],
                 "Predicted rating (SVD)": round(svd_pred_all[explain_user, mi], 2),
                 "Genre affinity (user's mean diff)": why,
                 "Content terms": ", ".join(
                     t for t in terms[np.argsort(-(X[mi].toarray().ravel() * prof))[:3]] if t)})
expl_df = pd.DataFrame(expl)
expl_df.to_csv(f"{OUT}/explanations_user{uid}.csv", index=False)
print(expl_df.to_string(index=False))

# ---- chart -----------------------------------------------------------------
try:
    import matplotlib
    matplotlib.use("Agg")
    import matplotlib.pyplot as plt
    fig, ax = plt.subplots(figsize=(7, 4))
    x = np.arange(len(results))
    ax.bar(x - 0.2, results["Precision@10"], 0.4, label="Precision@10")
    ax.bar(x + 0.2, results["Recall@10"], 0.4, label="Recall@10")
    ax.set_xticks(x)
    ax.set_xticklabels(results["Model"])
    ax.set_ylabel("Score")
    ax.set_title(f"Top-10 accuracy on {len(eval_users)} users (80/20 hold-out)")
    ax.legend()
    fig.tight_layout()
    fig.savefig(f"{OUT}/comparison.png", dpi=150)
    print("chart saved")
except Exception as e:
    print("chart skipped:", e)
```
